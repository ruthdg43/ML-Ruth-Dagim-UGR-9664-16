```python
df = pd.read_csv("classification_escalation.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
print("First 5 rows:")
display(df.head())

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()

print("\nMissing values:")
print(df.isnull().sum())

print("\nTarget distribution:")
print(df["escalated"].value_counts())

print("\nTarget percentages:")
print(df["escalated"].value_counts(normalize=True) * 100)
```

```python
from sklearn.model_selection import train_test_split

X = df.drop("escalated", axis=1)
y = df["escalated"]

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

print("Training size:", X_train.shape)
print("Testing size:", X_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())

print("\nTesting target distribution:")
print(y_test.value_counts())
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer

numeric_features = [
    "complexity_score",
    "team_experience_years",
    "customer_tenure_months"
]

categorical_features = ["priority", "channel"]

numeric_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

categorical_pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("encoder", OneHotEncoder(handle_unknown="ignore"))
])

preprocessor = ColumnTransformer([
    ("num", numeric_pipeline, numeric_features),
    ("cat", categorical_pipeline, categorical_features)
])

print("Preprocessing pipelines are ready!")
```

```python
from sklearn.dummy import DummyClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.discriminant_analysis import LinearDiscriminantAnalysis
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)

models = {
    "Baseline": DummyClassifier(strategy="most_frequent"),
    "Logistic Regression": Pipeline([
        ("preprocessing", preprocessor),
        ("model", LogisticRegression(max_iter=1000))
    ]),
    "LDA": Pipeline([
        ("preprocessing", preprocessor),
        ("model", LinearDiscriminantAnalysis())
    ]),
    "KNN (k=15)": Pipeline([
        ("preprocessing", preprocessor),
        ("model", KNeighborsClassifier(n_neighbors=15))
    ])
}

results = []

for name, model in models.items():
    model.fit(X_train, y_train)
    predictions = model.predict(X_test)

    if hasattr(model, "predict_proba"):
        probabilities = model.predict_proba(X_test)[:, 1]
        auc = roc_auc_score(y_test, probabilities)
    else:
        auc = np.nan

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(y_test, predictions, zero_division=0),
        "Recall": recall_score(y_test, predictions, zero_division=0),
        "F1": f1_score(y_test, predictions, zero_division=0),
        "ROC-AUC": auc
    })

results_df = pd.DataFrame(results)
display(results_df.round(3))
```

```python
from sklearn.metrics import confusion_matrix, ConfusionMatrixDisplay, roc_curve
import matplotlib.pyplot as plt

log_model = models["Logistic Regression"]
y_prob = log_model.predict_proba(X_test)[:, 1]
y_pred = (y_prob >= 0.5).astype(int)

cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

ConfusionMatrixDisplay(confusion_matrix=cm).plot()
plt.show()

fpr, tpr, thresholds = roc_curve(y_test, y_prob)

plt.plot(fpr, tpr, label="Logistic Regression")
plt.plot([0, 1], [0, 1], linestyle="--")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve")
plt.legend()
plt.show()
```

```python
from sklearn.metrics import roc_curve
import matplotlib.pyplot as plt

log_model = models["Logistic Regression"]
log_probs = log_model.predict_proba(X_test)[:, 1]

fpr, tpr, thresholds = roc_curve(y_test, log_probs)

plt.figure(figsize=(7, 5))
plt.plot(fpr, tpr, label="Logistic Regression (AUC = 0.811)")
plt.plot([0, 1], [0, 1], linestyle="--", label="Random classifier")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate (Recall)")
plt.title("ROC Curve - Logistic Regression")
plt.legend()
plt.grid()
plt.show()
```

```python
from sklearn.metrics import confusion_matrix

thresholds_to_test = [0.20, 0.30, 0.40, 0.50, 0.60, 0.70]

threshold_results = []

for threshold in thresholds_to_test:
    predictions = (log_probs >= threshold).astype(int)

    tn, fp, fn, tp = confusion_matrix(y_test, predictions).ravel()

    precision = precision_score(y_test, predictions, zero_division=0)
    recall = recall_score(y_test, predictions, zero_division=0)
    f1 = f1_score(y_test, predictions, zero_division=0)

    cost = 5 * fn + fp

    threshold_results.append({
        "Threshold": threshold,
        "Precision": precision,
        "Recall": recall,
        "F1": f1,
        "FP": fp,
        "FN": fn,
        "Cost (5FN + FP)": cost
    })

threshold_df = pd.DataFrame(threshold_results)
display(threshold_df.round(3))
```

```python
from sklearn.model_selection import StratifiedKFold, cross_val_score

k_values = [5, 9, 15, 21, 31]

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)

knn_cv_results = []

for k in k_values:
    knn_pipeline = Pipeline([
        ("preprocessing", preprocessor),
        ("model", KNeighborsClassifier(n_neighbors=k))
    ])

    scores = cross_val_score(
        knn_pipeline,
        X_train,
        y_train,
        cv=cv,
        scoring="f1"
    )

    knn_cv_results.append({
        "k": k,
        "Mean F1": scores.mean(),
        "Std F1": scores.std()
    })

knn_cv_df = pd.DataFrame(knn_cv_results)
display(knn_cv_df.round(3))
```

```python
best_knn = Pipeline([
    ("preprocessing", preprocessor),
    ("model", KNeighborsClassifier(n_neighbors=5))
])

best_knn.fit(X_train, y_train)

knn_predictions = best_knn.predict(X_test)
knn_probs = best_knn.predict_proba(X_test)[:, 1]

print("Best KNN (k=5) Test Results")
print("Accuracy:", accuracy_score(y_test, knn_predictions))
print("Precision:", precision_score(y_test, knn_predictions, zero_division=0))
print("Recall:", recall_score(y_test, knn_predictions, zero_division=0))
print("F1:", f1_score(y_test, knn_predictions, zero_division=0))
print("ROC-AUC:", roc_auc_score(y_test, knn_probs))
```

1. What are prior, likelihood, posterior, and evidence?
- Prior: Probability of escalation before looking at the customer's specific information. In our dataset, escalation prevalence is approximately 20.67%.
- Likelihood: How probable the observed customer information is if the case really is escalated.
- Posterior: Updated probability of escalation after observing the customer's information.
- Evidence: The overall probability of observing that customer information.
2. If the posterior probability is 0.30, and FP cost = 1, FN cost = 5, what should we predict?

Given:
- Probability of escalation = 0.30
- Probability of no escalation = 0.70
If we flag/escalate:
\[
Cost = 0.70 \times 1 = 0.70
\]

If we don't flag:
\[
Cost = 0.30 \times 5 = 1.50
\]

Therefore, we should predict escalation (1) because its expected cost is lower: 0.70 versus 1.50.

The decision threshold is:
\[
Threshold=\frac{FP\ Cost}{FP\ Cost+FN\ Cost}
=\frac{1}{1+5}=0.167
\]
